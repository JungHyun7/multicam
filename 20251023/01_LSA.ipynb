```python
# 라이브러리 로드
from konlpy.tag import Komoran
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.svm import SVC
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score, f1_score
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# 데이터 로드
df = pd.read_csv("../data/ratings_train.txt", sep='\t')
```

```python
# 데이터 결측치를 제외
df.dropna(inplace=True)
```

```python
# 독립 변수, 종속 변수
X = df['document'][:5000]
Y = df['label'][:5000]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, stratify=Y, random_state=42
)
```

```python
print(X_train.shape)
```

- 파이프라인 생성
    - 백터화(TF-IDF 방식)
        - 토큰화 함수(Komoran 형태소 방식) 포함
    -LSA
        - TruncatedSVD() 차원 축소
    - Sclaer
        - StandardScaler를 이용
    - Model
        - SVC() 모델

```python
# 토큰화 함수
komoran = Komoran()
allow_pos = ['NNP', 'NNG', 'VV', 'VA', 'MAG']

def tokenize(text):
    result = []
    for word, pos in komoran.pos(text):
        if pos in allow_pos:
            result.append(word)
    return result

# 백터화 객체 생성
vetorizer = TfidfVectorizer(
    # 토큰화 방식
    tokenizer= tokenize,
    lowercase=False,
    min_df=5,
    max_df = 0.8,
    ngram_range=(1, 1)
)
```

```python
# 차원 축소 객체 생성
svd = TruncatedSVD(
    n_components=200,
    random_state=42
)
```

```python
# StandardScaler
scaler = StandardScaler(with_mean=True)
```

```python
svc = SVC(
    kernel='linear',
    random_state=42,
    probability=False
)
```

```python
# 위에 4개 작업을 하나의 객체로 결합
pipe = Pipeline(
    [
        ('vector', vetorizer),
        ('lsa', svd),
        ('scaler', scaler),
        ('model', svc)
    ]
)
```

```python
# 최적의 매개변수를 찾기 위해 매개변수들을 지정
# pipe를 이용하게 되면 pipe step의 객체들의 key__parameter : [val1, val2]
grid_params = {
    "vector__ngram_range" : [(1, 1), (1, 2)],
    "lsa__n_components" : [100, 200],
    "model__C" : [1.0, 2.0]
}
```

```python
# GridSearchCV를 이용하여 각각 매개변수들을 대입하여 최적의 매개변수를 찾는다
grid = GridSearchCV(
    estimator=pipe,
    param_grid=grid_params,
    cv = 2,
    scoring='accuracy',
    verbose=1,
    #n_jobs=-1   # remote error 발생 시 주석처리
    
)
```

```python
grid.fit(X_train, Y_train)
```

```python
# 최적의 모델을 확인
print("Best Params : ", grid.best_params_)
# 최적의 모델의 정확도를 확인
print("Best Acc : ", round(grid.best_score_, 4))
```

```python
# 테스트 데이터를 이용하여 정화도 f1score를 확인
# 실무 -> 충분한 모델링 학습을 하고 실전의 데이터를 이용하여 정확도 및 f1score를 확인
# gird에서 최적의 모델을 변수에 저장
best_model = grid.best_estimator_
```

```python
# test 데이터셋을 이용하여 예측
best_pred = best_model.predict(X_test)
```

```python
# 예측 값과 실제 값을 이용하여 정확도 f1score을 확인
best_acc = accuracy_score(best_pred, Y_test)
best_f1 = f1_score(best_pred, Y_test)
print(f"정확도 : {round(best_acc, 4)}, F1 : {round(best_f1, 4)}")
```
