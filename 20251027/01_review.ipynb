```python
import os
from glob import glob
import pandas as pd
```

```python
# 특정 경로의 파일의 목록을 가져오는 기능
# os 라이브러리를 이용
os.listdir(
    './'
)
```

```python
# glob 이용
# 장점 : 파일의 경로와 파일의 이름을 하나의 리스트로 생성 특정 확장자만 선택해서 리스트로 생성이 가능
json_list = glob("./*.*.json")
```

```python
# json_list를 이용하여 하나의 데이터프레임으로 결합

new_list = []
new_df = pd.DataFrame()

for file_path in json_list[:5]:
    # print(file_path)
    df = pd.read_json(file_path)
    new_list.append(df)
    new_df = pd.concat([new_df, df], axis=0)
    # pirnt(df)
    # break

total_df = pd.concat(new_list, axis=0)
total_df.reset_index(drop=True, inplace=True)
new_df.reset_index(drop=True, inplace=True)
len(new_df)
len(total_df)
```

```python
# Aspects의 데이터를 하나로 합치고 새로운 데이터프레임을 생성
aspect_df = pd.DataFrame(sum(total_df['Aspects'], []))
```
